## 1. Создание и индексирование многомерных массивов

**NumPy** (Numerical Python) — базовая библиотека научных вычислений в Python. Её главный объект — **ndarray**: n-мерный массив элементов **одного типа**, хранящийся в непрерывном блоке памяти. Благодаря этому операции над массивами выполняются в десятки раз быстрее, чем над списками, а код получается компактнее.

Импортируют NumPy по стандартному соглашению:

```python
import numpy as np
```

````python
import numpy as np

a = np.array([1, 2, 3, 4, 5])            # одномерный массив из списка
m = np.array([[1, 2, 3],
              [4, 5, 6]])                # двумерный массив (матрица 2x3)

print(a)
print(m)
print('shape (форма):      ', m.shape)   # (2, 3): две строки, три столбца
print('ndim (размерность): ', m.ndim)    # 2
print('size (число элем.): ', m.size)    # 6
print('dtype (тип элем.):  ', m.dtype)   # int64
````

````python
# Типовые конструкторы массивов
print(np.arange(0, 10, 2))     # как range: от 0 до 10 с шагом 2
print(np.linspace(0, 1, 5))    # 5 равномерных точек на отрезке [0, 1]
print(np.zeros((2, 3)))        # матрица нулей 2x3
print(np.ones((3, 2)))         # матрица единиц 3x2
print(np.full((2, 2), 7.5))    # матрица, заполненная значением
print(np.eye(3))               # единичная матрица 3x3
````

````python
m = np.arange(12)                # [0, 1, ..., 11]
print(m.reshape(3, 4))           # та же последовательность, форма 3x4
print(m.reshape(4, 3).T)         # транспонированная матрица

f = m.astype(np.float64)         # новый массив с другим типом данных
print(f.dtype, f[:3])
````

### Индексирование и срезы

Синтаксис такой же, как у списков, но для многомерных массивов индексы перечисляются через запятую: `a[строка, столбец]`. Отрицательные индексы и срезы работают как обычно.

````python
scores = np.array([[5, 4, 3, 5],
                   [4, 4, 4, 3],
                   [3, 5, 4, 5]])    # оценки 3 студентов по 4 предметам

print(scores[1, 2])       # 2-я строка, 3-й столбец
print(scores[0])          # вся 1-я строка
print(scores[:, 0])       # весь 1-й столбец
print(scores[0:2, 1:3])   # строки 0-1, столбцы 1-2
print(scores[-1])         # последняя строка
print(scores[::-1])       # строки в обратном порядке
print(scores[0, ::2])     # первая строка, каждый второй столбец
````

**Важно: срез массива — это «представление» (view), а не копия.** Изменение среза меняет исходный массив (в отличие от срезов списков Python!). Если нужна независимая копия — используйте метод `.copy()`.

````python
row = scores[0]
row[0] = 99                 # изменили срез...
print(scores)               # ...и исходная матрица изменилась тоже

scores[0, 0] = 5            # вернём значение обратно
row_copy = scores[0].copy() # независимая копия
row_copy[0] = -1
print(scores[0], '| копия:', row_copy)
````

### Булевы маски и выбор по условию

Условие над массивом возвращает **булеву маску** — массив из `True` и `False`. Маской можно отфильтровать элементы или использовать её в агрегатах. Условия комбинируют операторами `&` (И), `|` (ИЛИ), `~` (НЕ) — каждое отдельное условие обязательно берут в скобки.

````python
data = np.array([15, 22, 8, 41, 6, 33, 27])

mask = data > 20
print(mask)                             # [False  True False  True False  True  True]
print(data[mask])                       # только элементы больше 20
print(data[data % 2 == 0])              # фильтрация в одну строку
print(data[(data > 10) & (data < 35)])  # И: скобки обязательны!
print(data[(data < 10) | (data > 40)])  # ИЛИ
print(data[~(data > 20)])               # НЕ

print(np.where(data > 20, 1, 0))        # замена по условию: маска -> числа
print(data[np.array([0, 3, 6])])        # fancy indexing: выбор по набору индексов
````

## 2. Векторизация вычислений и broadcasting

**Векторизация** — применение операции ко всему массиву сразу, без явного цикла. Операторы `+`, `-`, `*`, `/`, `**` работают **поэлементно**, а математические функции NumPy (`np.sqrt`, `np.exp`, `np.sin` и др.) обрабатывают массив целиком. Такие операции реализованы на языке C и выполняются на порядки быстрее циклов Python.

````python
prices = np.array([100.0, 250.0, 80.0, 320.0])
print(prices * 0.85)              # скидка 15% сразу всем товарам

q = np.array([1.0, 2.0, 3.0, 4.0])
print(q ** 2 + 2 * q + 1)         # выражение над всем массивом разом
print(np.sqrt(q))                 # универсальные функции (ufunc)
print(np.round(np.exp(q), 2))

# Для сравнения: список так не умеет
print([1, 2, 3] * 2)              # список повторится, а не умножится
````

### Broadcasting

**Broadcasting** — механизм, который позволяет NumPy выполнять операции над массивами **разных форм** без копирования данных. Правила:

1. Формы массивов сравниваются **с конца**.
2. Размерности совместимы, если они **равны** или одна из них равна **1**.
3. Массив с меньшим числом размерностей дополняется ведущими единицами.
4. Размерность, равная 1, «растягивается» до размера соответствующей размерности второго массива.

Пример: матрица `(3, 4)` плюс массив `(4,)` — массив дополняется до `(1, 4)` и складывается с каждой строкой матрицы.

````python
m = np.arange(1, 13).reshape(3, 4)
print('Матрица 3x4:')
print(m)

row = np.array([10, 20, 30, 40])        # форма (4,)
print('Матрица + строка (4,):')
print(m + row)                           # строка прибавилась к каждой строке

col = np.array([[1], [10], [100]])       # форма (3, 1)
print('Матрица + столбец (3,1):')
print(m + col)                           # столбец прибавился к каждому столбцу
````

````python
temps_c = np.array([15.0, 18.0, 21.0, 24.0])
temps_f = temps_c * 9 / 5 + 32           # перевод в Фаренгейты одной строкой
print(temps_f)

# Центрирование данных: вычитаем из каждого столбца его среднее
measurements = np.array([[10.0, 20.0, 30.0],
                         [12.0, 18.0, 34.0],
                         [11.0, 22.0, 29.0],
                         [ 9.0, 24.0, 27.0]])
col_means = measurements.mean(axis=0, keepdims=True)   # форма (1, 3)
print('Средние по столбцам:', col_means)
centered = measurements - col_means                    # broadcasting
print('Средние центрированных данных:', centered.mean(axis=0))   # почти нули
````

## 3. Агрегирующие и статистические функции NumPy

NumPy содержит готовый набор статистических функций. Большинство работает и как функция (`np.mean(a)`), и как метод массива (`a.mean()`).

````python
data = np.array([4, 8, 15, 16, 23, 42])

print('sum    ', data.sum())                             # сумма
print('mean   ', data.mean())                            # среднее
print('std    ', round(data.std(), 2))                   # стандартное отклонение
print('var    ', round(data.var(), 2))                   # дисперсия
print('min/max', data.min(), data.max())
print('argmax/argmin:', data.argmax(), data.argmin())    # ИНДЕКСЫ экстремумов
print('median ', np.median(data))                        # медиана (только функция)
print('p75    ', np.percentile(data, 75))                # 75-й перцентиль
print('cumsum ', data.cumsum())                          # накопленная сумма
````

### Агрегация по осям: параметр axis

Для многомерных массивов агрегаты считают вдоль **осей**. Правило запоминания: `axis=0` — «вдоль строк», в результате по одному значению на каждый столбец; `axis=1` — «вдоль столбцов», по одному значению на каждую строку.

````python
m = np.arange(1, 13).reshape(3, 4)
print('Матрица:')
print(m)
print('Сумма всех элементов:   ', m.sum())
print('Суммы по столбцам (0): ', m.sum(axis=0))    # 4 значения - по числу столбцов
print('Суммы по строкам (1):  ', m.sum(axis=1))    # 3 значения - по числу строк
print('Средние по строкам:    ', m.mean(axis=1))
print('Максимумы по столбцам: ', m.max(axis=0))
````

### Пропуски: nan-функции

`np.nan` (not a number) — стандартное обозначение пропущенного значения в аналитических данных. Обычные функции возвращают nan, если в данных есть хотя бы один пропуск, поэтому существуют варианты с префиксом `nan`, игнорирующие пропуски.

````python
readings = np.array([1.0, 2.5, np.nan, 4.0, 3.5])

print(readings.mean())           # nan: один пропуск испортил весь результат
print(np.nanmean(readings))      # среднее без учёта пропусков
print(np.nanmax(readings))
print(np.isnan(readings))        # маска пропусков
print('Пропусков:', np.isnan(readings).sum())
````

## 4. Сравнение циклических и векторизованных операций

Проверим на практике, насколько векторизация быстрее цикла. Для замеров в Jupyter есть магия `%timeit` (многократно повторяет код и усредняет время), а для самостоятельных замеров — модуль `time` и функция `perf_counter`.

````python
import numpy as np
from time import perf_counter

rng = np.random.default_rng(42)          # генератор случайных чисел с «семенем»
n = 1_000_000
x = rng.random(n)                        # массив из n случайных чисел на [0, 1)
x_list = x.tolist()                      # те же числа в виде списка Python

# Вариант 1: классический цикл по списку
start = perf_counter()
total = 0.0
for v in x_list:
    total += v * v
loop_time = perf_counter() - start

# Вариант 2: векторизованная операция NumPy
start = perf_counter()
result = np.sum(x * x)
vector_time = perf_counter() - start

print(f'Цикл по списку:     {loop_time * 1000:8.1f} мс')
print(f'Векторизация NumPy: {vector_time * 1000:8.2f} мс')
print(f'Ускорение: примерно в {loop_time / vector_time:.0f} раз')
print('Результаты совпадают:', np.isclose(total, result))
````

````python
# Магия %timeit: повторяет замер много раз и даёт устойчивую оценку
%timeit sum(v * v for v in x_list)   # сумма квадратов по списку (генератор)
%timeit np.sum(x * x)                # векторизованный вариант
````

**Почему векторизация быстрее:**

- на каждой итерации цикла Python интерпретатор проверяет типы и создаёт объекты — это дорого;
- NumPy выполняет операцию один раз над всем массивом скомпилированным кодом на C;
- элементы массива лежат в непрерывной памяти и имеют один тип — нет накладных расходов на отдельные объекты.

**Когда цикл всё же уместен:** небольшие данные; логика, которую нельзя выразить поэлементными операциями (например, значение зависит от предыдущего шага); шаги с побочными эффектами (печать, запись в файл).

**Правило:** если операцию можно записать формулой над массивом — записывайте её векторно.

### Задание 5. Измеряем скорость

1. Создайте массив `x` из `1_000_000` случайных чисел (`default_rng`, метод `random`) и его списочную копию (`tolist`).
2. Напишите функцию `normalize_loop(values)`: циклом строит и возвращает список, где каждое значение поделено на максимум.
3. Векторный вариант — одна строка: `x / x.max()` (для массива).
4. Измерьте время обоих вариантов (`perf_counter` или `%timeit`).
5. Выведите, во сколько раз векторизация быстрее.
6. Убедитесь, что результаты совпадают: `np.allclose`.

````python
# Решение задания 5
rng=np.random.default_rng()
x = rng.random(1_000_000) 
tolist=x.tolist()

def normalize_loop(values):
    maxx=max(values)
    new=[]
    for x in values:
        new.append(x/maxx)
    return new

start=perf_counter()
loop_r=normalize_loop(tolist)
loop_time = perf_counter() - start

start=perf_counter()
vector_r=x / x.max()
vector_time = perf_counter() - start

print("Время цикла:", loop_time)
print("Время векторизации:", vector_time)
print("Векторизация быстрее в",loop_time / vector_time,"раз")
print('Результаты совпадают:', np.allclose(vector_r, loop_r))
````

5. `anomaly(temps)` — отклонение каждой дневной температуры от среднего по её месяцу (broadcasting с `keepdims=True`);
6. `climate_report(temps)` — возвращает словарь: средняя температура за год, самый тёплый и самый холодный месяцы, число дней с температурой ниже 0.

**Требования:** у каждой функции — докстринг; весь числовой код — без циклов, только векторные операции; печать результатов — только в конце конвейера.

````python
import numpy as np

# Данные для итоговой задачи
rng = np.random.default_rng(2024)
month_index = np.arange(12).reshape(12, 1)
seasonal = 10 + 15 * np.sin((month_index - 3) / 12 * 2 * np.pi)
TEMPS = seasonal + rng.normal(0, 3, size=(12, 30))

# Решение итоговой задачи
def generate_temperatures(seed=2024):
    """Создаёт массив температур размером 12x30 по заданной модели."""
    rng = np.random.default_rng(2024)
    month_index = np.arange(12).reshape(12, 1)
    seasonal = 10 + 15 * np.sin((month_index - 3) / 12 * 2 * np.pi)
    TEMPS = seasonal + rng.normal(0, 3, size=(12, 30))
    return TEMPS

def monthly_mean(temps):
    """массив средних температур по месяцам, округлённых до одного знака"""
    return np.round(temps.mean(axis=1), 1)

def warmest_month(temps):
    """возвращает номер самого тёплого месяца (от 1 до 12) и его среднюю температуру"""
    return np.argmax(temps.mean(axis=1))+1, round(max(temps.mean(axis=1)), 1)


def hot_days(temps, threshold=25):
    """число дней в году с температурой выше `threshold`"""
    return (temps>threshold).sum()

def anomaly(temps):
    """отклонение каждой дневной температуры от среднего по её месяцу"""
    return temps-temps.mean(axis=1, keepdims=True)

def climate_report(temps):
    """возвращает словарь: средняя температура за год, самый тёплый и самый холодный месяцы, число дней с температурой ниже 0."""
    a=temps.mean(axis=1)
    b=np.argmax(a)
    c=np.argmin(a)
    return {
        "Средняя температура за год:": round(temps.mean(), 1),
        "Самый тёплый месяц:": b+1,
        "Средняя температура тёплого месяца:": round(a[b], 1),
        "Самый холодный месяц:": c+1,
        "Средняя температура холодного месяца:": round(a[c], 1),
        "Число дней с температоруй ниже 0:": (temps<0).sum()
    }
    
temps = generate_temperatures()
print("Средние температуры по месяцам:", monthly_mean(temps), '\n')
print("Номер самого тёплого месяца и его средняя температура:", warmest_month(temps)[0], 'Месяц', '  ', warmest_month(temps)[1], 'Средняя температура', '\n')
print("Число дней в году с температурой выше 25:", hot_days(temps), 'дней', '\n')
print("Отчет:")
for key, value in  climate_report(temps).items():
    print(key, value)
print("")
print("Отклонение каждой дневной температуры от среднего по её месяцу:", '\n', anomaly(temps))
````

## Контрольные вопросы

1. Чем массив NumPy принципиально отличается от списка Python? Назовите минимум три отличия.

Ответ: 1)NumPy содержит элементы одного типа, а список может содержать разные типы.
       2)NumPy быстрее обрабатывает большие объёмы числовых данных.
       3)NumPy намного удобнее для работы с многомерными данными и матрицами.

2. Что означают записи `a[1]`, `a[1, 2]` и `a[1][2]` для двумерного массива? Какой из вариантов предпочтительнее и почему?

Ответ:a[1] — берём вторую строку; a[1, 2] — берём элемент второй строки и третьего столбца; a[1][2] — сначала берём вторую строку, а потом из неё третий элемент(два этапа вместо одного). Вариант a[1, 2] лучше т.к. он быстрее.

3. Почему срез массива — это представление (view)? Как получить независимую копию?

Ответ: Срез NumPy по умолчанию создаёт представление исходного массива, а не новый массив.Изменение среза меняет исходный массив (в отличие от срезов списков Python). Если нужна независимая копия — используется метод `.copy()`.

4. Сформулируйте правила broadcasting. Совместимы ли формы `(4, 1)` и `(1, 3)`? Какая форма получится у результата?

Ответ:1) Формы массивов сравниваются **с конца**.
      2) Размерности совместимы, если они **равны** или одна из них равна **1**.
      3) Массив с меньшим числом размерностей дополняется ведущими единицами.
      4) Размерность, равная 1, «растягивается» до размера соответствующей размерности второго массива.
    Совместимы, результат:(4, 3)

5. Что вернёт `m.sum(axis=0)` для матрицы `m` формы `(3, 4)`? А `m.mean(axis=1)`?

Ответ: 1)Вернет 1 строку с 4 столбцами с суммой по столбцам (1, 4). 
       2)Вернет 1 столбец с 3 строками с средним значением по строкам (3, 1)

6. Зачем скобки в выражении `(data > 0) & (data < 10)`? Что произойдёт без них?

Ответ: Скобки нужны, чтобы сначала отдельно получить две логические маски, а потом объединить их через &. Без скобок Python неправильно обработает приоритет операторов и можно получить ошибку или неправильное выражение.

7. Чем `argmax` отличается от `max`? В каких задачах нужен именно `argmax`?

Ответ: 'argmax' возвращает индекс максимального значения, 'max' возвращает максимальное значение. Argmax используется при поиске, например, самой дорогой вещи в магазине(с max мы будем знать только её стоимость).

8. Почему векторизованные операции NumPy быстрее циклов Python?

Ответ:Математические функции NumPy обрабатывают массив целиком. Такие операции реализованы на языке C и выполняются на порядки быстрее циклов Python, потому что циклы выполняются поэлементно.

9. Когда цикл всё же предпочтительнее векторизации?

Ответ: Когда цикл всё же уместен: небольшие данные; логика, которую нельзя выразить поэлементными операциями (например, значение зависит от предыдущего шага); шаги с побочными эффектами (печать, запись в файл).

10. Чем `np.nanmean` отличается от `mean`? Когда это критично?

Ответ:`np.nan` (not a number) — стандартное обозначение пропущенного значения в аналитических данных. Обычные функции возвращают nan, если в данных есть хотя бы один пропуск, поэтому существуют варианты с префиксом `nan`, игнорирующие пропуски. Это критично при работе с реальными данными, где могут быть пропуски. Если использовать обычный mean, пропущенное значение может сделать весь результат NaN.
